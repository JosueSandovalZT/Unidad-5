# Ejercicio 1 - Interpolacion segmentada (Spline)

```python
import numpy as np # importa numpy para manejo de datos numericos
from scipy.interpolate import CubicSpline # importa el metodo de spline cubico

x = np.array([0,1,2,3]) # Define los puntos en el eje x
y = np.array([1,3,2,5]) # Define los valores correspondientes

spline = CubicSpline(x,y) # Construye el spline cubico

x_interp = np.linspace(0,3,100) # Genera 100 puntos para evaluacion

y_interp = spline(x_interp) # Evalua el spline en los puntos de interpolacion

print("valores interpolados : ",y_interp) # Muestra los valores
```
