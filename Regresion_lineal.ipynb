# Ejercicio 2 - Regresion lineal

```python
import numpy as np  # Importa NumPy

x = np.array([1,2,3,4,5])  # Datos de entrada
y = np.array([2,4,5,4,5])  # Datos observados

coef = np.polyfit(x,y,1)   # Ajusta un polinomio de grado 1 (recta)

m = coef[0]   # Pendiente de la recta
b = coef[1]   # Intersección con el eje y

y_pred = m*x + b  # Calcula los valores predichos por el modelo

print("Pendiente:",m)      # Muestra la pendiente
print("Intercepto:",b)     # Muestra el término independiente
print("Predicción:",y_pred) # Muestra los valores estimados
```
